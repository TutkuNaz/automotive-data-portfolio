# NHTSA vehicle-complaint starter
Fetch public complaint records for one vehicle and inspect the most frequently reported components. Complaint counts are descriptive signals, not failure rates: they are not normalized by the number of vehicles in service and reflect voluntary reporting behavior.

```python
import collections, json, urllib.parse, urllib.request
make, model, model_year = 'HONDA', 'CIVIC', 2020
query = urllib.parse.urlencode({'make': make, 'model': model, 'modelYear': model_year})
url = f'https://api.nhtsa.gov/complaints/complaintsByVehicle?{query}'
request = urllib.request.Request(url, headers={'User-Agent': 'automotive-open-data-notebook/1.0'})
with urllib.request.urlopen(request, timeout=30) as response:
    complaints = json.load(response).get('results', [])
len(complaints)
```

```python
component_counts = collections.Counter(row.get('components') or 'UNKNOWN' for row in complaints)
component_counts.most_common(10)
```

```python
def flagged(field):
    return sum(str(row.get(field, '')).upper() in {'Y', 'YES', 'TRUE'} for row in complaints)
{'vehicle': f'{model_year} {make} {model}', 'complaints': len(complaints), 'crash_reports': flagged('crash'), 'fire_reports': flagged('fire')}
```
